## Entrenamiento y corriendo un modelo de regresión lineal usando Scikit-Learn

```python
# Librerias a usar
#=====================================================
import matplotlib.pyplot as plt
import numpy as np 
import pandas as pd 
from sklearn.linear_model import LinearRegression
```

```python
# Datos en la tabla (DataFrame)
#==================================================
datos = {
    "Country": ["Turkey", "Hungray", "France", "United States", "New Zealand", "Autralia", "Denmark"],
    "GDP per capita (USD)": [28384, 31008, 42026, 60236, 42404, 48698, 55938],
    "Life satisfaction": [5.5, 5.6, 6.5, 6.9, 7.3, 7.3, 7.6]
}

tabla = pd.DataFrame(datos)
print("-------------------------- Tabla --------------------------")
print (tabla)
```

```python
# Descargamos y preparamos los datos
#===================================================
x = tabla[["GDP per capita (USD)"]].values 
y = tabla[["Life satisfaction"]].values
```

```python
# Visualización de los datos
#===================================================
tabla.plot(kind='scatter', grid=True, x="GDP per capita (USD)", y="Life satisfaction")
# Ajustar los ejes
plt.axis([23_500, 62_500, 4, 9])
# Mostrar la gráfica
plt.show()
```

```python
# iniciamos y entrenamos el modelo de regresión lineal 
#============================================================================
#seleccionamos el modelo
model = LinearRegression()
# Entrenamos el modelo
model.fit(x,y)
```

```python
# Hacemos una predicción (Aquí se toma el ejemplo USD) 
X_new =[[37_655.2]] 
print(model.predict(X_new))
```

## Probando un modelo de regresión K-narest regression

```python
# Remplazamos el modelo de regresión lineal con k-nearest neightbors
#====================================================================
from sklearn.neighbors import KNeighborsRegressor
```

```python
# Descargamos y preparamos los datos
#===================================================
x = tabla[["GDP per capita (USD)"]].values 
y = tabla[["Life satisfaction"]].values
```

```python
# Visualización de los datos
#===================================================
tabla.plot(kind='scatter', grid=True, x="GDP per capita (USD)", y="Life satisfaction")
# Ajustar los ejes
plt.axis([23_500, 62_500, 4, 9])
# Mostrar la gráfica
plt.show()
```

```python
# iniciamos y entrenamos el modelo de regresión lineal 
#============================================================================
#seleccionamos el modelo
model = KNeighborsRegressor(n_neighbors=3)
# Entrenamos el modelo
model.fit(x,y)
```

```python
# Hacemos una predicción (Aquí se toma el ejemplo USD) 
X_new =[[37_655.2]] 
print(model.predict(X_new))
```
